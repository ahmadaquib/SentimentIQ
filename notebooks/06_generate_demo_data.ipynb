# Demo Data Generation Notebook

This notebook reads sample texts from `data/sample_inputs.csv` and runs them through the SentimentIQ prediction pipeline (`backend/predictor.py`), populating `data/predictions.csv` to showcase full analytics in the Streamlit dashboard.

```python
import os
import sys
sys.path.append('..')

import pandas as pd
from backend.predictor import predict_text
from backend.csv_logger import log_prediction, get_analytics_summary

df = pd.read_csv('../data/sample_inputs.csv')
print(f'Loaded {len(df)} sample texts for demo data generation.')
```

```python
for idx, row in df.iterrows():
    t = str(row['text']).strip()
    if t:
        pred = predict_text(t)
        log_prediction(pred)

print('Demo predictions generated and logged!')
```

```python
summary = get_analytics_summary()
print('=== Current Analytics Summary ===')
print(summary)
```
